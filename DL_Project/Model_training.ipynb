```python
df = pd.read_csv("cleaned_health_data.csv")
df.head()
```

```python
df_model = df.drop(columns=['Country'])
```

```python
X = df_model.drop(columns=['Adult_Mortality', 'infant_deaths'])

y_adult = df_model['Adult_Mortality']
y_infant = df_model['infant_deaths']
```

```python
X_train_a, X_test_a, y_train_a, y_test_a = train_test_split(
    X, y_adult,
    test_size=0.2,
    random_state=42
)
```

```python
rf_adult = RandomForestRegressor(
    n_estimators=200,
    random_state=42
)

rf_adult.fit(X_train_a, y_train_a)
```

```python
xgb_adult = XGBRegressor(
    n_estimators=200,
    learning_rate=0.05,
    max_depth=6,
    random_state=42
)

xgb_adult.fit(X_train_a, y_train_a)
```

```python
for name, model in {
    "Random Forest": rf_adult,
    "XGBoost": xgb_adult
}.items():

    preds = model.predict(X_test_a)

    print(f"\n{name} - Adult Mortality")
    print("MAE:", mean_absolute_error(y_test_a, preds))
    print("RMSE:", np.sqrt(mean_squared_error(y_test_a, preds)))
    print("R2:", r2_score(y_test_a, preds))
```

```python
X_train_i, X_test_i, y_train_i, y_test_i = train_test_split(
    X, y_infant,
    test_size=0.2,
    random_state=42
)
```

```python
rf_infant = RandomForestRegressor(
    n_estimators=200,
    random_state=42
)

rf_infant.fit(X_train_i, y_train_i)
```

```python
xgb_infant = XGBRegressor(
    n_estimators=200,
    learning_rate=0.05,
    max_depth=6,
    random_state=42
)

xgb_infant.fit(X_train_i, y_train_i)
```

```python
for name, model in {
    "Random Forest": rf_infant,
    "XGBoost": xgb_infant
}.items():

    preds = model.predict(X_test_i)

    print(f"\n{name} - Infant Deaths")
    print("MAE:", mean_absolute_error(y_test_i, preds))
    print("RMSE:", np.sqrt(mean_squared_error(y_test_i, preds)))
    print("R2:", r2_score(y_test_i, preds))
```

```python
joblib.dump(xgb_adult, "adult_mortality_model.pkl")
joblib.dump(xgb_infant, "infant_deaths_model.pkl")

print("Models saved successfully.")
```
